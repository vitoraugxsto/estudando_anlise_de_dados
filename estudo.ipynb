```python
import pandas as pd

data = pd.read_csv('dataframe.csv')
```

```python
data.head()
```

```python
data.rename(columns=
  {
    'Date' : 'Data',
    'Region' : 'Regiao',
    'Product Name' : 'Nome do Produto',
    'Units Sold' : 'Unidades Vendidas',
    'Unit Price' : 'Preco Unitario',
    'Total Revenue' : 'Receita Total',
    'Payment Method' : 'Metodo de Pagamento',
    'Product Category' : 'Categoria do Produto'

}, inplace = True
)
```

```python
data['Receita Esperada'] = data['Unidades Vendidas'] * data['Preco Unitario']
data.head()
```

```python
qtd_credito = (data['Metodo de Pagamento'] == 'Credit Card').sum()
qtd_debit = (data['Metodo de Pagamento'] == 'Debit Card').sum()
qtd_paypal = (data['Metodo de Pagamento'] == 'PayPal').sum()
```

```python
qtd_creditodois = data.query('`Metodo de Pagamento` == "Credit Card"')sum()
qtd_creditodois
```

```python
data_Pagamentos =  {
    'Cartao de Credito' : [qtd_credito],
    'Cartao de Debito' : [qtd_debit],
    'Paypal' : [qtd_paypal]
}
```
